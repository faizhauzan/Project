## **Model Development**
I chose to use the **Random Forest Classifier** because my dataset is extensive and diverse, comprising a total of 736,282 rows and 24 columns. This model is particularly well-suited for handling large and complex datasets due to its ability to process a significant volume of data efficiently while maintaining high accuracy. Additionally, the **Random Forest Classifier** is robust to outliers and performs well even when the data distribution is imbalanced, ensuring reliable predictions across various scenarios. Its ensemble nature further reduces overfitting, making it an ideal choice for my classification task.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
```

```python
df['Iron_Concentrate_Class'] = df['% Iron Concentrate'].apply(lambda x: 1 if x > 64 else 0)

features = ['% Iron Feed', '% Silica Feed', 'Starch Flow', 'Amina Flow',
            'Ore Pulp Flow', 'Ore Pulp pH', 'Ore Pulp Density',
            'Average Air Flow', 'Average Froth Level']
```

In this code, I’m preparing the data for a classification model to detect high-grade iron ore versus low-grade iron ore. First, I created a new column called `Iron_Concentrate_Class`. This column is a binary label where :     
- If the % Iron Concentrate is greater than 64, the ore is classified as high grade (1).
- If it’s 64 or lower, the ore is classified as low grade (0).

Next, I defined a list of features that will be used to predict the grade of the iron ore. These features include parameters like `% Iron Feed`, `% Silica Feed`, `Starch Flow`, `Amina Flow`, and others that are related to the ore processing process. These features will serve as the input to the model, which will learn how these variables relate to the classification of high-grade or low-grade ore. ***The goal is to train a model that can predict whether the iron ore is high or low grade based on these process variables.***

```python
X = df[features]
y = df['Iron_Concentrate_Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = RandomForestClassifier(n_estimators=100, random_state=42)
```

This code splits the data into features (X) and the target variable (y), which is whether the iron ore is high or low grade. It then splits the data into training and testing sets, with ***30% for testing and 70% for training***. A `RandomForestClassifier` model is created with 100 decision trees. The model will be trained using the training data and tested on the test data to predict the iron ore quality.

```python
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

The model is trained using the training data `(X_train and y_train)` with the `fit()` method. After training, it makes predictions on the test data` (X_test)` using the `predict()` method, generating predicted values `(y_pred)` for the iron ore classification.

```python
print(classification_report(y_test, y_pred))

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")

importances = model.feature_importances_
print(f"Feature Importances: {importances}")
```

The model performs excellently, with an accuracy of 99.10%. It has high precision and recall for both classes, meaning it correctly identifies both high and low-grade iron ore with minimal errors. The f1-scores of 0.97 for class 0 (low-grade) and 0.99 for class 1 (high-grade) further confirm the model's effectiveness.

In terms of feature importance, the most influential features are `Ore Pulp pH` and `Amina Flow` followed by `Starch Flow` and `Iron Feed`. These features play a significant role in determining whether the ore is high or low grade.
